# Cell 5 — What is Deep Learning?

```markdown
<div style="background-color:#F3E5F5;padding:15px;border-left:6px solid #9C27B0;border-radius:8px;">

<h2 style="color:#6A1B9A;">🧠 What is Deep Learning?</h2>

</div>

### Professional Definition

**Deep Learning is a subfield of Machine Learning that uses neural networks with multiple computational layers to learn hierarchical representations of data by optimizing parameterized functions using gradient-based optimization.**

A deep neural network learns a mapping:

$$
f_{\theta}: X \rightarrow Y
$$

where:

- $X$ = input data
- $Y$ = target/output
- $\theta$ = learnable parameters
- $f_{\theta}$ = neural network

The objective is to find parameters $\theta$ that minimize a loss function:

$$
\theta^* = \arg\min_{\theta} L(f_{\theta}(X),Y)
$$

### Core idea

Instead of explicitly programming every rule, the network learns parameters from examples.

<div style="background-color:#E3F2FD;padding:15px;border-left:6px solid #1976D2;border-radius:8px;">

<h2 style="color:#0D47A1;">🏗️ Why is it called "Deep" Learning?</h2>

</div>

The word **deep** primarily refers to the presence of multiple learnable computational layers between the input and output.

Example:

  ↓

  ↓

  ↓

```markdown
<div style="background-color:#E0F7FA;padding:15px;border-left:6px solid #00ACC1;border-radius:8px;">

<h2 style="color:#006064;">⚙️ Artificial Neuron</h2>

</div>

An artificial neuron computes a weighted combination of its inputs followed by a nonlinear activation function.

$$
z = \sum_{i=1}^{n} w_i x_i + b
$$

or:

$$
z = W^TX+b
$$

The output is:

$$
y = \phi(z)
$$

where:

- $x_i$ = input
- $w_i$ = weight
- $b$ = bias
- $z$ = pre-activation
- $\phi$ = activation function
- $y$ = neuron output

### Computational flow

```text

x₂ ──w₂──┤
x₃ ──w₃──┤──> Σ + b ──> Activation ──> Output

# Cell 12 — Forward Propagation

```markdown
<h2 style="color:#1565C0;">➡️ Forward Propagation</h2>

Forward propagation calculates the network's output from input data.

For layer $l$:

$$
z^{(l)} = W^{(l)}a^{(l-1)} + b^{(l)}
$$

Then:

$$
a^{(l)} = \phi^{(l)}(z^{(l)})
$$

For a multi-layer network:

$$
X
\rightarrow
Z_1
\rightarrow
A_1
\rightarrow
Z_2
\rightarrow
A_2
\rightarrow
...
\rightarrow
\hat{Y}
$$

where:

- $X$ = input
- $W$ = weights
- $b$ = bias
- $A$ = activation
- $\hat{Y}$ = prediction

<h2 style="color:#C62828;">📉 Loss Function</h2>

A loss function measures the difference between the predicted output and the true target.

$$
Loss = L(y,\hat{y})
$$

For regression, a common loss is Mean Squared Error:

$$
MSE =
\frac{1}{n}
\sum_{i=1}^{n}
(y_i-\hat{y}_i)^2
$$

For binary classification:

$$
Binary\ Cross\ Entropy =
-\frac{1}{N}
\sum_{i=1}^{N}
[y_i\log(\hat y_i)
+
(1-y_i)\log(1-\hat y_i)]
$$

The objective is:

$$
\min_{\theta} L(\theta)
$$

Lower loss generally means that the model's predictions are closer to the training targets.

<div style="background-color:#FFEBEE;padding:15px;border-left:6px solid #E53935;border-radius:8px;">

<h2 style="color:#B71C1C;">⬅️ Backpropagation</h2>

</div>

Backpropagation computes gradients of the loss with respect to the network parameters using the chain rule of calculus.

For a parameter $\theta$:

$$
\frac{\partial L}{\partial \theta}
$$

The chain rule allows the gradient to propagate backward through multiple layers.

```text
Output Error
     ↑
     │
Layer L
     ↑
Layer L-1
     ↑
Layer L-2
     ↑
     .
     .
Input Layer

# Cell 15 — Gradient Descent

```markdown
<h2 style="color:#6A1B9A;">📐 Gradient Descent</h2>

Parameters are updated using:

$$
\theta_{t+1}
=
\theta_t
-
\eta
\nabla_{\theta}L(\theta_t)
$$

where:

- $\theta$ = model parameters
- $\eta$ = learning rate
- $\nabla L$ = gradient of loss

A very small learning rate:

# Cell 24 — Deep Learning Mathematical View

```markdown
<div style="background-color:#E8EAF6;padding:15px;border-left:6px solid #3F51B5;border-radius:8px;">

<h2 style="color:#283593;">📐 Deep Learning as Function Approximation</h2>

</div>

A neural network approximates a function:

$$
\hat{y}=f_{\theta}(x)
$$

For a network with $L$ layers:

$$
h_1 = \sigma(W_1x+b_1)
$$

$$
h_2 = \sigma(W_2h_1+b_2)
$$

$$
...
$$

$$
\hat{y}=f(W_Lh_{L-1}+b_L)
$$

Training attempts to find:

$$
\theta^*
=
\arg\min_{\theta}
\mathbb{E}_{(x,y)\sim D}
[L(f_{\theta}(x),y)]
$$

This is the fundamental mathematical formulation of supervised Deep Learning.

````python
import numpy as np

# Input features
x = np.array([2.0, 3.0, 4.0])

# Weights
w = np.array([0.5, 0.8, -0.2])

# Bias
b = 1.0

# Weighted sum
z = np.dot(x, w) + b

print("Weighted Sum:", z)
````
