## 1. What is a Data Type?

A **data type (dtype)** tells us what kind of values a column contains and how Pandas should store/process those values.

Common Pandas dtypes include:

| dtype | Meaning | Example |
|---|---|---|
| `int64` | Integer numbers | `10`, `25`, `100` |
| `float64` | Decimal/floating-point numbers | `10.5`, `99.99` |
| `object` | General Python objects; commonly strings/text | `'Pune'`, `'Male'`, `'M'` |
| `bool` | Boolean values | `True`, `False` |
| `datetime64[ns]` | Date/time values | `2026-08-18` |
| `category` | Categorical values | `'B'`, `'M'` |

⚠️ **Important:** `object` does not always mean "string". It is a generic dtype that can hold Python objects, although in typical tabular datasets it commonly represents text/string columns.

## 2. `int64` — Integer Data

`int64` represents **whole numbers without decimal values**.

### Examples
```python
age = 25
experience = 3
employees = 100
```

Typical DataFrame columns:
- Age
- Number of children
- Quantity
- Employee count
- IDs when they are truly numeric and used numerically

### Example
```python
import pandas as pd

df = pd.DataFrame({
    'age': [22, 25, 30, 35],
    'experience': [0, 2, 5, 10]
})

print(df.dtypes)
```

Expected result:
```text
age           int64
experience    int64
dtype: object
```

## 3. `float64` — Decimal Data

`float64` represents **numbers that can contain decimal values**.

### Examples
```python
salary = 45000.50
height = 5.11
rating = 4.5
temperature = 31.7
```

Typical DataFrame columns:
- Salary
- Height / Weight
- Price
- Percentage
- Measurement values
- ML features such as radius, area, probability, etc.

## 4. `object` — Text / General Python Objects

`object` is commonly seen when a column contains **strings/text**.

### Example
```python
name = 'Kunal'
city = 'Pune'
gender = 'Male'
diagnosis = 'M'
```

Pandas may represent these columns as `object`.

### Common examples
| Column | Example values | Typical dtype |
|---|---|---|
| Name | `Kunal`, `Amit` | `object` |
| City | `Pune`, `Nashik` | `object` |
| Gender | `Male`, `Female` | `object` |
| Diagnosis | `B`, `M` | `object` |
| Department | `IT`, `HR` | `object` |

### ⚠️ Important interview point

A column such as `salary` can accidentally become `object` if the data is stored as strings:

```text
'45000'
'50000'
'55000'
```

These look numeric to a human, but they are strings to Python/Pandas.

````python
import pandas as pd

df = pd.DataFrame({
    'age': [22, 25, 30, 35],
    'salary': [35000.50, 45000.75, 55000.00, 65000.25],
    'city': ['Pune', 'Nashik', 'Mumbai', 'Pune'],
    'experience': [0, 2, 5, 10]
})

df
````

````python
# Check the data type of every column
df.dtypes
````

## 5. `df.dtypes` vs `df.info()`

### `df.dtypes`
Shows the dtype of each column.

```python
df.dtypes
```

### `df.info()`
Provides a broader summary:
- Number of rows
- Number of columns
- Column names
- Non-null counts
- Data types
- Approximate memory usage

```python
df.info()
```

💡 **Interview answer:** Use `dtypes` when you mainly need column types; use `info()` when you want a quick structural and missing-value overview.

## 6. Why does Pandas use `int64` and `float64`?

`64` indicates the typical number of bits used to represent the numeric value in NumPy/Pandas.

- `int64` → 64-bit integer
- `float64` → 64-bit floating-point number

Do not confuse the dtype with the number of decimal places. `float64` does **not** mean "64 decimal places".

## 7. Type Conversion with `astype()`

Use `astype()` when you know the conversion is valid.

```python
df['age'] = df['age'].astype(float)
```

Convert a numeric column to integer:

```python
df['age'] = df['age'].astype('int64')
```

Convert text categories to string:

```python
df['city'] = df['city'].astype(str)
```

⚠️ `astype()` can raise an error if values cannot be converted.

## 8. `pd.to_numeric()` — Safer Numeric Conversion

Suppose a salary column contains:

```text
'45000'
'50000'
'Not Available'
'60000'
```

Use:

```python
df['salary'] = pd.to_numeric(df['salary'], errors='coerce')
```

`errors='coerce'` converts invalid values to `NaN`, which can then be handled during data cleaning.

### Important distinction
- `astype()` → direct conversion; invalid values can raise an error.
- `pd.to_numeric(..., errors='coerce')` → invalid numeric values become `NaN`.

## 9. Real-World Example: Numeric Values Stored as `object`

```python
df = pd.DataFrame({
    'salary': ['35000', '45000', '55000', '65000']
})

print(df.dtypes)
```

The salary column may be `object` because the values are strings.

Convert it:

```python
df['salary'] = pd.to_numeric(df['salary'])
```

Now Pandas can perform numerical operations correctly.

## 10. Why Data Types Matter in Machine Learning

Most ML algorithms require numerical input.

Example:

```text
Age        → int64       ✅
Salary     → float64     ✅
City       → object      ⚠️
Gender     → object      ⚠️
```

`City` and `Gender` cannot usually be passed directly into a numerical ML algorithm. Categorical features need suitable encoding, such as **One-Hot Encoding** or another appropriate encoding strategy.

Example:

```text
Gender
Male
Female
Male
```

can be transformed using one-hot encoding into numerical columns such as:

```text
Gender_Male
Gender_Female
```

# 🧠 Interview Questions — Frequently Asked

## Basic Questions

### Q1. What is `int64` in Pandas?
**Answer:** `int64` represents 64-bit integer values without decimal parts.

### Q2. What is `float64`?
**Answer:** `float64` represents 64-bit floating-point numbers, which can contain decimal values.

### Q3. What is `object` dtype?
**Answer:** `object` is a general Pandas/NumPy dtype for Python objects. In ordinary tabular datasets, it most commonly represents strings/text.

### Q4. How do you check column data types?
**Answer:**
```python
df.dtypes
```
or
```python
df.info()
```

## Intermediate Interview Questions

### Q5. Why can a numeric column have dtype `object`?
**Answer:** Because the numeric-looking values may actually be stored as strings, or the column may contain mixed/incompatible Python objects.

### Q6. How do you convert an object column to numeric?
```python
df['salary'] = pd.to_numeric(df['salary'], errors='coerce')
```

### Q7. Difference between `astype()` and `pd.to_numeric()`?
**Answer:** `astype()` performs an explicit dtype conversion. `pd.to_numeric()` is designed specifically for numeric conversion and supports error handling such as `errors='coerce'`.

### Q8. What does `errors='coerce'` do?
**Answer:** Values that cannot be converted to numbers are replaced with `NaN` instead of causing the conversion to fail.

### Q9. Can `object` columns be used directly in ML models?
**Answer:** Usually not when the model expects numerical features. Text/categorical columns generally need preprocessing/encoding first.

### Q10. Why should you inspect dtypes before ML preprocessing?
**Answer:** Incorrect dtypes can cause calculation errors, prevent models from accepting data, produce incorrect analysis, and hide data-quality problems.

## 🔥 Scenario-Based Interview Questions

### Q11. Your `salary` column is `object`, but it contains numbers. What will you do?
**Expected approach:** Inspect the values first, clean symbols/commas if required, then use `pd.to_numeric()`.

Example:
```python
df['salary'] = df['salary'].str.replace(',', '', regex=False)
df['salary'] = pd.to_numeric(df['salary'], errors='coerce')
```

### Q12. Your ML model gives an error saying it cannot convert strings to float. What would you check?
**Answer:** Check `df.dtypes`, identify object/string columns, inspect their unique values, and encode or remove inappropriate columns.

```python
print(df.dtypes)
print(df.select_dtypes(include='object').columns)
```

### Q13. An ID column is `int64`. Should you always treat it as a numerical feature?
**Answer:** No. An ID can be numerically stored but semantically be only an identifier. Whether to use it as a feature depends on the problem and data-generation process. Often it is excluded from model features.

### Q14. What happens if a categorical column contains numbers like `1, 2, 3`?
**Answer:** Numeric storage does not automatically mean the variable is continuous. You must understand the meaning of the values. If `1, 2, 3` represent categories, treat them as categorical rather than assuming mathematical distance.

### Q15. Why is dtype inspection part of EDA?
**Answer:** Dtypes reveal how the dataset is represented and help detect data-quality issues before analysis, visualization, feature engineering, and model training.
