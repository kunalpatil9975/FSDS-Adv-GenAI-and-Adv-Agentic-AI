````python
# Data Visualization Library
import seaborn as sns

# Data Manipulation and Analysis Library
import pandas as pd

# Plotting Library
import matplotlib.pyplot as plt

# Numerical Computing Library
import numpy as np
````

````python
# Create an Empty DataFrame
df = pd.DataFrame()
````

````python
# Import 'X' constant from the re (Regular Expression) module
from re import X
````

````python
# Create a Pandas Series named x
x = pd.Series([12, 25, 68, 42, 113])

# Create a Pandas Series named y
y = pd.Series([11, 29, 58, 121, 100])
````

````python
# Add the x Series as a new column 'x' in the DataFrame
df['x'] = x

# Add the y Series as a new column 'y' in the DataFrame
df['y'] = y
````

````python
# Display the DataFrame
df
````

````python
# Create a figure with 1 row and 2 columns of subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 3))

# Plot the original scatter plot
ax1.scatter(df['x'], df['y'])

# Plot the scaled scatter plot (x and y multiplied by 2)
ax2.scatter(df['x'] * 2, df['y'] * 2)

# Set the title with the covariance of the original data
ax1.set_title("Covariance - " + str(np.cov(df['x'], df['y'])[0, 1]))

# Set the title with the covariance of the scaled data
ax2.set_title("Covariance - " + str(np.cov(df['x'] * 2, df['y'] * 2)[0, 1]))
````

## Observation

- The left scatter plot represents the original dataset with a covariance of **1148.75**.
- The right scatter plot represents the scaled dataset, where both **X** and **Y** values are multiplied by **2**.
- The pattern and direction of the data points remain the same in both plots.
- The covariance increases from **1148.75** to **4595.0** because covariance is affected by the scale of the variables.
- This demonstrates that **covariance is scale-dependent**, while the relationship between the variables remains unchanged.

````python
# Print the covariance between the original X and Y variables
print(np.cov(df['x'], df['y'])[0,1])

# Print the covariance after scaling both X and Y by a factor of 2
print(np.cov(df['x'] * 2, df['y'] * 2)[0,1])
````

````python
# Create a figure with 1 row and 3 columns of subplots
fig, ax = plt.subplots(1, 3, figsize=(15, 3))

# Plot X vs X on the first subplot
ax[0].scatter(df['x'], df['x'])

# Plot X vs Y on the second subplot
ax[1].scatter(df['x'], df['y'])

# Plot scaled data (2X vs 2Y) on the third subplot
ax[2].scatter(df['x'] * 2, df['y'] * 2)

# Display the covariance of X with itself (Variance of X)
ax[0].set_title("Covariance - " + str(np.cov(df['x'], df['x'])[0,1]))

# Display the covariance between X and Y
ax[1].set_title("Covariance - " + str(np.cov(df['x'], df['y'])[0,1]))

# Display the covariance after scaling both X and Y by 2
ax[2].set_title("Covariance - " + str(np.cov(df['x'] * 2, df['y'] * 2)[0,1]))
````

## Observation

- The first plot shows **X vs X**, where the covariance is equal to the **variance of X**.
- The second plot shows the relationship between **X and Y** with their original covariance.
- The third plot shows the scaled data (**2X vs 2Y**), where the covariance becomes **4 times larger**.
- The pattern and direction of the data points remain the same in all three plots.
- This demonstrates that **covariance is affected by the scale of the variables**, whereas the relationship between the variables remains unchanged.

## Summary

- Covariance measures the direction of the relationship between two variables.
- A positive covariance indicates that both variables tend to increase together, while a negative covariance indicates an inverse relationship.
- Covariance is **scale-dependent**, meaning its value changes when the variables are scaled.
- Multiplying both variables by **2** increases the covariance by **4 times**.
- Although the covariance value changes after scaling, the relationship and pattern between the variables remain unchanged.
- Covariance between a variable and itself is equal to its **variance**.
- Due to its dependence on scale, **correlation** is generally preferred for comparing relationships between variables.

### Correlation

````python
# Create a figure with 1 row and 2 columns of subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 3))

# Plot X vs X on the first subplot
ax1.scatter(df['x'], df['x'])

# Plot X vs Y on the second subplot
ax2.scatter(df['x'], df['y'])

# Display the correlation between X and itself
ax1.set_title("Correlation - " + str(df['x'].corr(df['x'])))

# Display the correlation between X and Y
ax2.set_title("Correlation - " + str(df['x'].corr(df['y'])))
````

## Observation

- The first scatter plot shows **X vs X**, and the correlation is **1.0** because a variable is perfectly correlated with itself.
- The second scatter plot shows the relationship between **X and Y**.
- The correlation value lies between **-1 and +1**.
- A positive correlation indicates that both variables tend to increase together.
- Unlike covariance, **correlation is scale-independent**, so multiplying the data by a constant does not change the correlation value.

````python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 3))

# Plot scatterplots on each axes
ax1.scatter(df['x'], df['y'])
ax2.scatter(df['x']*2, df['y'])

ax1.set_title("Correlation - " + str(df['x'].corr(df['y'])))
ax2.set_title("Correlation - " + str((df['x']*2).corr(df['y']*2)))
````

## Observation

- The first scatter plot shows **X vs X**, and the correlation is **1.0** because a variable is perfectly correlated with itself.
- The second scatter plot shows the relationship between **X and Y**.
- The correlation value lies between **-1 and +1**.
- A positive correlation indicates that both variables tend to increase together.
- Unlike covariance, **correlation is scale-independent**, so multiplying the data by a constant does not change the correlation value.

## Summary

- Correlation measures the **strength and direction** of the relationship between two variables.
- Its value always ranges from **-1 to +1**.
- A correlation of **+1** indicates a perfect positive relationship.
- A correlation of **0** indicates no linear relationship.
- Correlation is **scale-independent**, making it more suitable than covariance for comparing relationships between variables.
