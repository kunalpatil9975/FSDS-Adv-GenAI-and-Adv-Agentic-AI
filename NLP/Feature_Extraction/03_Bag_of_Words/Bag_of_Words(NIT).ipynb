## 📖 Definition

**Bag of Words (BoW)** is a **feature extraction technique** in Natural Language Processing (NLP) that converts text into a **numerical representation** by counting the frequency of each word in a document.

> **It ignores grammar, word order, and context.**

---

## 🎯 Why BoW is Used?

- Converts text into numbers
- Makes text understandable for Machine Learning models
- Creates a vocabulary of unique words
- Counts the occurrence of each word
- Simple and fast implementation

## 📝 Interview Definition

### 📖 Bag of Words (BoW)

- **Bag of Words (BoW)** is a text representation technique used in NLP.
- It converts text documents into a **numerical vector (word count matrix)**.
- Each column represents a **unique word** from the vocabulary.
- Each row represents a **document**, and each value is the **frequency (count)** of that word.
- BoW ignores the **order and grammar** of words and only keeps their occurrence count.

## ❌ Drawbacks (Disadvantages) of Bag of Words (BoW)

### 1. Ignores Word Order
- BoW does not consider the sequence of words.
- Example:
  - "Dog bites man"
  - "Man bites dog"
- Both may have similar word counts but different meanings.

---

### 2. Ignores Context
- BoW cannot understand the meaning of words in different contexts.
- Example:
  - "Bank of river"
  - "Bank account"

---

### 3. Creates Sparse Matrix
- Most values in the BoW matrix are **0**, making it memory inefficient.

---

### 4. Large Vocabulary
- Every unique word becomes a feature.
- More unique words = More columns = Higher memory usage.

---

### 5. Cannot Handle Synonyms
- Similar words are treated as different words.
- Example:
  - Happy
  - Joyful
  - Glad

---

### 6. Cannot Handle Polysemy
- A word with multiple meanings is treated as the same word.
- Example:
  - "Bat" (Animal)
  - "Bat" (Sports equipment)

---

### 7. No Semantic Meaning
- BoW counts words but does not understand their relationships or meanings.

---

### 8. Sensitive to Noise
- Misspelled words, punctuation, and unnecessary words increase the vocabulary.

---

### 9. High Dimensionality
- A large dataset creates thousands of features, increasing computational cost.

---

### 10. Poor Performance on Large Datasets
- Accuracy decreases when context and semantics are important.

---

**Drawbacks of Bag of Words:**
- Ignores word order

- Large vocabulary
- Cannot understand synonyms
- Cannot understand multiple meanings of words
- No semantic understanding
- High dimensionality

# 15. What is a Sparse Matrix?

### Answer

A **Sparse Matrix** is a matrix where **most values are zero**.

### Example

| Doc | NLP | Python | AI | Java |
|----|----|----|----|----|
| D1 |1|0|0|0|
| D2 |0|1|0|0|
| D3 |0|0|1|0|

Most values are **0**, so this is called a Sparse Matrix.

### Interview Point

BoW and TF-IDF usually produce sparse matrices.

---

# 16. Why does BoW create a Sparse Matrix?

### Answer

Because every document contains only a few words from the complete vocabulary.

Example

Vocabulary Size = 20,000 words

Document contains only 50 words.

Remaining 19,950 values become **0**.

### Interview Point

Large vocabulary ⇒ More zeros ⇒ Sparse Matrix.

# 18. What is the Curse of Dimensionality?

### Answer

The Curse of Dimensionality occurs when the number of features becomes very large, making Machine Learning models slower and less efficient.

### Problems

- More memory usage
- Longer training time
- Increased overfitting
- Lower accuracy
- Difficult visualization

### Interview Point

BoW may suffer from the Curse of Dimensionality due to its large vocabulary.

## 21. Can BoW understand synonyms?

### Answer:
**No.** BoW treats synonyms as completely different words.

**Example:**

```

```

Although these words have similar meanings, BoW considers them as three separate features.

**Interview Point:**
BoW cannot capture semantic similarity between words.

## 22. Can BoW understand polysemy?

### Answer:
**No.** BoW cannot distinguish between multiple meanings of the same word.

**Example:**

```
I deposited money in the bank.
I sat on the river bank.
```

Here, **bank** has two different meanings, but BoW treats both as the same word.

**Interview Point:**
BoW ignores context.

---

## 23. What happens when a new word appears during testing?

### Answer:
If a word was **not present in the training vocabulary**, CountVectorizer ignores it.

**Example**

Training Vocabulary

```
love
python
nlp
```

Test Sentence

```
I love AI
```

Word **AI** is ignored because it is not in the vocabulary.

**Interview Point:**
BoW cannot handle unseen (Out-of-Vocabulary) words.

---

## 24. How does CountVectorizer handle unseen words?

### Answer:
CountVectorizer only transforms words that already exist in the learned vocabulary.

Unknown words are ignored.

**Example**

```
Vocabulary
love
python
nlp
```

Input

```
I love AI
```

Output

```

python = 0
nlp = 0
```

The word **AI** is discarded.

**Interview Point:**
Vocabulary is fixed after `fit()`.

## 30. Difference between BoW and BERT.

| Bag of Words | BERT |
|--------------|------|

BoW only counts words, whereas BERT generates context-aware embeddings where the meaning of a word depends on the surrounding words.

**CountVectorizer parameters** control how text is preprocessed, tokenized, filtered, and converted into a Bag of Words feature matrix.

# <center><span style="color:#8E24AA;">📚 ngram_range Parameter in CountVectorizer</span></center>

---

# 📖 Definition

`ngram_range` is a **parameter** of `CountVectorizer()` that specifies the **minimum (min_n)** and **maximum (max_n)** number of consecutive words (N-grams) to extract as features.

### Syntax

```python
CountVectorizer(ngram_range=(min_n, max_n))
```

- **min_n** → Minimum number of words in an N-gram.
- **max_n** → Maximum number of words in an N-gram.

---

# General Syntax

```python
CountVectorizer(
    ngram_range=(min_n, max_n)
)
```

---

# What is an N-gram?

An **N-gram** is a sequence of **N consecutive words**.

Example sentence:

```text
I love Machine Learning
```

| N | Name | Output |
|---|------|--------|
| 1 | Unigram | I, love, Machine, Learning |
| 2 | Bigram | I love, love Machine, Machine Learning |
| 3 | Trigram | I love Machine, love Machine Learning |
| 4 | Four-gram | I love Machine Learning |

---

# Why use ngram_range?

- Captures word sequences.
- Preserves some word order.
- Improves text classification.
- Better than using only single words in many tasks.

---

# Advantages

- Captures phrase-level information.
- Improves sentiment analysis.
- Better context than BoW with only unigrams.
- Easy to implement.

---

# Drawbacks

- Increases vocabulary size.
- Creates larger sparse matrices.
- Requires more memory.
- Slower training for higher-order N-grams.

---

# Applications

- Sentiment Analysis
- Spam Detection
- Text Classification
- Search Engines
- Document Classification
- Language Modeling
