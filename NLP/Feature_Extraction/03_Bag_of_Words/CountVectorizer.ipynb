## 📖 Definition

**CountVectorizer()** is a **Scikit-learn** class used to convert a collection of text documents into a **Bag of Words (BoW) matrix** by counting the frequency of each word.

> **Simple Definition:** CountVectorizer converts text into numerical vectors by counting how many times each word appears in a document.

---

## 📦 Import

```python
from sklearn.feature_extraction.text import CountVectorizer

# Create CountVectorizer Object
cv = CountVectorizer()
```

---

## 🎯 Technique Used

- **Bag of Words (BoW)**
- **N-Grams** (using `ngram_range` parameter)

---

## ⚙️ Internal Working of CountVectorizer()

```text
Input Documents
       │
       ▼
Convert Text to Lowercase
       │
       ▼
Tokenization
       │
       ▼
Remove Stopwords (Optional)
       │
       ▼
Apply N-Grams (Optional)
       │
       ▼
Create Vocabulary
       │
       ▼
Assign Index to Each Word
       │
       ▼
Count Word Frequency
       │
       ▼
Create Sparse Matrix
       │
       ▼
Bag of Words (BoW) Matrix
```

## Most Important Parameters

| Parameter | Description | Example |
|-----------|-------------|---------|
| `stop_words` | Removes common stopwords | `stop_words='english'` |
| `lowercase` | Converts text to lowercase | `lowercase=True` |
| `max_features` | Keeps only the top N frequent words | `max_features=100` |
| `min_df` | Ignores words appearing in fewer than N documents | `min_df=2` |
| `max_df` | Ignores words appearing in more than a specified number/percentage of documents | `max_df=0.8` |
| `binary` | Uses 0/1 instead of word frequency | `binary=True` |
| `ngram_range` | Creates Unigrams, Bigrams, Trigrams | `ngram_range=(1,2)` |
| `token_pattern` | Regex pattern used for tokenization | `token_pattern=r'(?u)\b\w+\b'` |
| `analyzer` | Type of features to extract | `'word'`, `'char'`, `'char_wb'` |
| `vocabulary` | Uses a custom vocabulary | `vocabulary=['python','nlp']` |
| `dtype` | Data type of the output matrix | `dtype=int` |
| `encoding` | Text encoding | `encoding='utf-8'` |
| `decode_error` | Action when decoding fails | `'ignore'` |
| `strip_accents` | Removes accents from characters | `'ascii'` or `'unicode'` |
| `preprocessor` | Custom preprocessing function | `preprocessor=my_func` |
| `tokenizer` | Custom tokenizer function | `tokenizer=my_tokenizer` |

````python
documents = [
    "I love NLP",
    "I love Python",
    "Python is useful for NLP"
]
````

````python
from sklearn.feature_extraction.text import CountVectorizer

# Sample Documents
documents = [
    "I love NLP",
    "I love Python",
    "Python is useful for NLP"
]

# Create CountVectorizer Object
cv = CountVectorizer()

# Learn Vocabulary
cv.fit(documents)
print("Vocabulary Learned Successfully")


````

````python


# Convert Documents into Bag of Words Matrix
bow = cv.transform(documents)
print("\nBag of Words Matrix:")
print(bow.toarray())

````

````python
# Learn Vocabulary + Convert Documents
bow = cv.fit_transform(documents)
print("\nfit_transform() Output:")
print(bow.toarray())

````

````python
# Display Vocabulary
print("\nFeature Names:")
print(cv.get_feature_names_out())

````

````python
# Display Vocabulary Dictionary
print("\nVocabulary Dictionary:")
print(cv.vocabulary_)


````

````python
# Convert Matrix Back to Words
print("\nInverse Transform:")
print(cv.inverse_transform(bow))


````

````python
# Display All Parameters
print("\nParameters:")
print(cv.get_params())


````

````python
# Change Parameter Values
cv.set_params(stop_words='english')
print("\nUpdated CountVectorizer:")
print(cv)
````
