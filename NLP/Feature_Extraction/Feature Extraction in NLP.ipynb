# <center><span style="color:#8E24AA;">📚 Feature Extraction in NLP</span></center>

---

## 📖 Definition

**Feature Extraction** is the process of converting **text data into numerical features (vectors)** so that Machine Learning and Deep Learning models can understand and process the text.

> **In simple words:** Feature Extraction converts words into numbers.

---

## 🎯 Why Feature Extraction?

Machine Learning algorithms cannot understand text directly.

Example:

```
I love NLP
```

Machine Learning understands only numbers.

```
[0.25, 0.61, 0.12]
```

Therefore, Feature Extraction converts text into numerical vectors.

---

## 🛠️ Workflow

```
Raw Text
    │
    ▼
Text Preprocessing
(Lowercase, Stopwords, etc.)
    │
    ▼
Tokenization
    │
    ▼
Feature Extraction
    │
    ▼
Numerical Vectors
    │
    ▼
Machine Learning Model
```

# <center><span style="color:#8E24AA;">📚 Feature Extraction Techniques in NLP (Interview Notes)</span></center>

---

## 1. Bag of Words (BoW)

**Definition:** Converts text into numerical vectors by counting the frequency of each word.

**Interview Point:** Ignores grammar, context, and word order.

---

## 2. Count Vectorizer

**Definition:** A Scikit-learn implementation of Bag of Words that converts text into a word-count matrix.

**Interview Point:** Automatically creates vocabulary and count matrix.

---

## 3. TF-IDF (Term Frequency–Inverse Document Frequency)

**Definition:** Assigns importance to words based on their frequency in a document and rarity across all documents.

**Interview Point:** Reduces the importance of common words like "the", "is", "and".

---

## 4. N-Grams

**Definition:** Represents text using sequences of N consecutive words.

**Examples:**
- Unigram (1 word)
- Bigram (2 words)
- Trigram (3 words)

**Interview Point:** Captures partial word order.

---

## 5. One-Hot Encoding (OHE)

**Definition:** Represents each unique word as a binary vector.

**Interview Point:** Simple but creates very high-dimensional sparse vectors.

---

## 6. Hashing Vectorizer

**Definition:** Uses a hash function to convert words into numerical features without storing a vocabulary.

**Interview Point:** Memory efficient and suitable for very large datasets.

---

# <center><span style="color:#1565C0;">Embedding Methods</span></center>

---

## 7. Word2Vec

**Definition:** Learns dense vector representations of words based on surrounding words.

**Interview Point:** Similar words have similar vector representations.

---

## 8. GloVe (Global Vectors for Word Representation)

**Definition:** Generates word embeddings using global word co-occurrence statistics.

**Interview Point:** Uses statistics from the entire corpus.

---

## 9. FastText

**Definition:** Represents words using character n-grams.

**Interview Point:** Handles unknown and misspelled words better than Word2Vec.

---

## 10. Doc2Vec

**Definition:** Learns dense vector representations for complete documents or paragraphs.

**Interview Point:** Used for document classification and similarity.

---

## 11. BERT Embeddings
(Bidirectional Encoder Representations from Transformers)

**Definition:** Generates context-aware word embeddings using Transformer architecture.

**Interview Point:** The same word can have different embeddings depending on the sentence context.

---

## 12. Sentence-BERT (SBERT)

**Definition:** Generates embeddings for complete sentences.

**Interview Point:** Mainly used for semantic search, sentence similarity, and retrieval tasks.
