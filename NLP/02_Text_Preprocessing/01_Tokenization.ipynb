## Definition

**Tokenization** is the process of breaking raw text into smaller meaningful units called **tokens**. These tokens can represent **sentences, words, characters, or subwords**.

It is one of the **first and most important preprocessing steps** in every **Natural Language Processing (NLP)** pipeline because machine learning and deep learning models cannot directly understand raw text.

---

## Why is Tokenization Important?

- Converts unstructured text into structured data.
- Makes text understandable for NLP algorithms.
- Required before feature extraction (Bag of Words, TF-IDF, Word2Vec).
- Improves NLP model performance.
- Reduces text processing complexity.

<h1 align="center" style="color:purple; font-size:28px;">
    Sentence Tokenization
</h1>

## Definition

**Sentence Tokenization** is the process of splitting a paragraph or document into individual sentences. It is also known as **Sentence Segmentation**.

Sentence Tokenization is one of the first preprocessing steps in Natural Language Processing (NLP). It helps NLP models understand sentence boundaries before performing further analysis.

---

## Why is Sentence Tokenization Important?

- Splits large paragraphs into individual sentences.
- Makes text easier to process and analyze.
- Improves the accuracy of NLP applications.

````python
# Input Text
text = """

India is a land of rich cultural diversity, where people from different religions, languages, and traditions live together in harmony. 
The country is known for its historical monuments, colorful festivals, delicious cuisine, and vibrant heritage. 
India has made remarkable progress in science, technology, healthcare, education, and space research over the past few decades. 
With the rapid growth of Artificial Intelligence, Machine Learning, and Natural Language Processing, many industries are adopting smart technologies to improve efficiency and decision-making. 
The unity, resilience, and innovation of the Indian people continue to inspire the world and contribute to the nation's development.
"""
````

````python
print(text)
````

````python
len(text)
````

## Basic Example

````python
# Import Sentence Tokenizer
from nltk.tokenize import sent_tokenize

# Split paragraph into sentences
sentences = sent_tokenize(text)

# Print all sentences
print(sentences)
````

````python
# Print each sentence separately

for sentence in sentences:
    print(sentence)
````

````python
# Print sentence number with each sentence

for i, sentence in enumerate(sentences, start=1):
    print(f"{i}. {sentence}")
````

###  Count Sentences

````python
# Count total number of sentences

print("Total Sentences:", len(sentences))
````

### Store in List

````python
# Sentence Tokenizer returns a Python list

print(type(sentences))
print(sentences)
````

## Indexing

````python
# Access sentences using index

print("First Sentence :", sentences[0])

print("Second Sentence:", sentences[1])

print("Last Sentence  :", sentences[-1])
````

````python
# Store sentences inside Pandas DataFrame

import pandas as pd

df = pd.DataFrame(sentences, columns=["Sentence"])

print(df)
````

````python
# Count characters in each sentence

for sentence in sentences:
    print(sentence)
    print("Characters:", len(sentence))
    print("-"*40)
````

<h1 align="center" style="color:purple; font-size:28px;">
    Word Tokenization
</h1>

### **Definition**
Word Tokenization is the process of splitting a sentence into individual words (tokens).

### **Why is Word Tokenization Needed?**
- Computers cannot understand raw text directly.
- Converts a sentence into individual words.
- First step in NLP preprocessing.
- Used for Stopword Removal, Stemming, Lemmatization, POS Tagging, Feature Extraction, and Machine Learning models.

### Difference Between Sentence Tokenization and Word Tokenization

| **Sentence Tokenization** | **Word Tokenization** |
|---------------------------|-----------------------|
| Splits a paragraph into sentences | Splits a sentence into words |
| Output is a list of sentences | Output is a list of words/tokens |
| Uses `.`, `?`, `!` to detect sentence boundaries   | Uses spaces and punctuation to split words |
| Also called **Sentence Segmentation** | Also called **Word Segmentation** |

````python
# Import Word Tokenizer
from nltk.tokenize import word_tokenize

# Convert text into word tokens
words = word_tokenize(text)

# Print all word tokens
print(words)
````

````python
# Total Number of Word Tokens

print("Total Words:", len(words))
````

````python
# Print each word separately

for word in words:
    print(word)
````

````python
# Print each word with its number

for i, word in enumerate(words, start=1):
    print(f"{i}. {word}")
````

````python
# First 20 Words
print(words[:20])
````

````python
# Last 20 Words

print(words[-20:])
````

````python
# Total Unique Words

unique_words = set(words)

print(unique_words)
print("Total Unique Words:", len(unique_words))
````

````python
# Word Frequency
from collections import Counter

frequency = Counter(words)

print(frequency)
````

````python
#Join Words Back into Text
original_text = " ".join(words)

print(original_text)
````

````python
# Access words using index

print("First Word :", words[0])

print("Second Word:", words[1])

print("Last Word  :", words[-1])
````

</h1>

### **Definition**
#### WordPunctTokenizer is a rule-based tokenizer in NLTK that splits text into words and punctuation as separate tokens.

### **Why is WordPunctTokenizer Needed?**
- Separates words and punctuation into individual tokens.
- Useful for text preprocessing and NLP tasks.

````python
from nltk.tokenize import WordPunctTokenizer

punct = """
Hello! My name is Kunal, and I'm learning NLP.
Do you know Python 3.12? It's amazing: fast, simple, and powerful.
Today, I scored 95%; however, my friend scored only 88/100.
Email me at kunal@example.com (or visit https://example.com) - Thanks!
"""
````

````python
# Print all tokens
print("\nTokens:\n")
for i, token in enumerate(tokens, start=1):
    print(f"{i}. {token}")
````

<h1 style="text-align:center; color:#99567A;">
Treebank Tokenization
</h1>

### **Definition**
Treebank Tokenization is a rule-based tokenization technique that follows the Penn Treebank rules to split text into words, punctuation, and contractions while preserving the grammatical structure of a sentence.

### **Why is Treebank Tokenization Needed?**
- Produces consistent and standardized tokens.
- Separates punctuation correctly.
- Splits contractions into meaningful tokens.
- Preserves grammatical structure.
- Improves NLP preprocessing accuracy.

### **Rules Used**
- Penn Treebank Rules
- Punctuation Rules
- Contraction Rules

### **Key Features**
- Rule-based tokenization.
- Standard English tokenization.
- Separates punctuation into individual tokens.
- Splits contractions correctly.
- Produces NLP-ready tokens.

### **Applications**
- POS Tagging
- Parsing
- Named Entity Recognition (NER)
- Machine Translation
- Text Classification

## Rules Used in TreebankWordTokenizer

### 1. Penn Treebank Rules

**What is it?**

Penn Treebank Rules are a set of standard English tokenization rules developed by the **University of Pennsylvania**. These rules define how words, punctuation, quotes, brackets, and contractions should be split into tokens.

**Why are they used?**

- To produce consistent tokenization.
- To preserve grammatical structure.
- To prepare text for NLP tasks such as POS tagging and parsing.

---

### 2. Punctuation Rules

**What is it?**

Punctuation Rules define how punctuation marks should be separated from words during tokenization.

**Why are they used?**

- To separate punctuation into individual tokens.
- To improve text preprocessing.
- To make grammatical analysis easier.

**Example**

```text
Input : Hello, World!
```

```python
Output : ['Hello', ',', 'World', '!']
```

---

### 3. Contraction Rules

**What is it?**

Contraction Rules define how contracted English words should be split into meaningful tokens.

**Why are they used?**

- To separate the base word and contraction.
- To improve grammatical understanding.
- To increase POS tagging and parsing accuracy.

**Example**

```text
Input : I can't swim.
```

```python
Output : ['I', 'ca', "n't", 'swim', '.']
```

**Treebank Tokenization** is the **concept** that defines how English text should be tokenized according to the **Penn Treebank Rules**.

**TreebankWordTokenizer** is the **NLTK implementation** of that concept. It applies the **Penn Treebank Rules** in Python to split text into **words**, **punctuation**, and **contractions**.

### **Why is MWE Tokenization Needed?**

- Treats related words as one token.
- Improves NLP preprocessing.
- Useful for phrase detection and information retrieval.

### **Rules Used**

- Phrase Matching Rules
- Dictionary-based Rules

### **Key Features**

- Combines multiple words into one token.
- Uses predefined word expressions.
- No training data required.
- Produces meaningful phrase tokens.

### **Advantages**
- Preserves phrase meaning.
- Easy to customize.
- Improves text preprocessing.
- Useful for search engines and NLP tasks.
