# <span style="color:#6A1B5B;">1.Byte Pair Encoding (BPE)</span>

</div>

### **Definition**
#### Byte Pair Encoding (BPE) is a modern subword tokenization algorithm that builds a vocabulary by repeatedly merging the most frequently occurring pair of characters or subwords into a single token.

### **Why is BPE Needed?**
- Handles unknown (Out-of-Vocabulary) words.
- Reduces vocabulary size.
- Represents rare words effectively.
- Improves Transformer and LLM performance.
- Creates reusable subword vocabulary.

### **How Does BPE Work?**
1. Split each word into individual characters.
2. Count the frequency of adjacent character pairs.
3. Merge the most frequent pair into a new token.
4. Repeat the process until the desired vocabulary size is reached.

### **Key Features**
- Subword-based tokenization.
- Frequency-based merging.
- Handles unseen words.
- Smaller vocabulary.
- Widely used in modern NLP.

### **Advantages**
- Handles unknown words.
- Reduces vocabulary size.
- Improves model efficiency.
- Better text representation.

### **Limitations**
- Vocabulary must be trained.
- Common words may still be split.
- Merge rules depend on the training corpus.

### **Applications**
- GPT-2
- RoBERTa
- CLIP
- Transformer Models
- Large Language Models (LLMs)

## Why Do We Use Hugging Face for BPE?

**Byte Pair Encoding (BPE)** is a **subword tokenization algorithm**, not a built-in tokenizer like those available in NLTK.

Unlike NLTK tokenizers (such as `word_tokenize()` or `TreebankWordTokenizer`), **BPE requires a trained vocabulary and merge rules**. These are learned from a large text corpus during the tokenizer training process.

Therefore, BPE is commonly implemented using libraries such as:

- **Hugging Face Transformers**
- **Hugging Face Tokenizers**

These libraries provide **pre-trained BPE tokenizers** used by modern Transformer and Large Language Models (LLMs), such as **GPT-2** and **RoBERTa**.

### Why Can't We Use NLTK?

- NLTK provides **traditional rule-based tokenizers**.
- It does **not** include implementations of modern subword tokenization algorithms such as:
  - Byte Pair Encoding (BPE)
  - WordPiece
  - SentencePiece
  - Unigram Language Model

### Why Use Hugging Face?

- Provides pre-trained BPE tokenizers.
- Uses the same tokenization as modern AI models.
- Supports GPT, RoBERTa, LLaMA, Mistral, and many other Transformer models.
- Easy to use with just a few lines of code.

# <span style="color:#6A1B9A;">2.WordPiece Tokenization</span>

</div>

### **Definition**
#### WordPiece is a modern subword tokenization algorithm that splits words into meaningful subword units using a probability-based vocabulary. It is widely used in BERT and other Transformer models.

### **Why is WordPiece Needed?**
- Handles unknown (Out-of-Vocabulary) words.
- Reduces vocabulary size.
- Represents rare words effectively.
- Improves Transformer model performance.
- Creates reusable subword vocabulary.

### **How Does WordPiece Work?**
1. Starts with a predefined vocabulary.
2. Splits unknown words into smaller subwords.
3. Selects the most probable subword sequence.
4. Uses **##** to indicate that a token continues from the previous token.

### **Key Features**
- Subword-based tokenization.
- Probability-based algorithm.
- Uses predefined vocabulary.
- Handles unknown words.
- Smaller vocabulary.

### **Advantages**
- Better handling of rare words.
- Reduces vocabulary size.
- Improves language understanding.
- Efficient for Transformer models.

### **Limitations**
- Requires a trained vocabulary.
- More complex than word tokenization.
- Depends on vocabulary quality.

### **Applications**
- BERT
- DistilBERT
- ELECTRA
- Transformer Models

````python
from transformers import AutoTokenizer
from collections import Counter

# Load BERT WordPiece Tokenizer
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

# Input Text
text = "Artificial Intelligence and Machine Learning are changing the world."

# Tokenize
tokens = tokenizer.tokenize(text)

# Print Tokens
print(tokens)

# Total Tokens
print("Total Tokens:", len(tokens))

# Unique Tokens
print("Unique Tokens:", len(set(tokens)))

# Token Frequency
frequency = Counter(tokens)

print(frequency)

# Top 10 Frequent Tokens
print(frequency.most_common(10))
````

<h1 style="text-align:center; color:#F57C00;">
SentencePiece Tokenization
</h1>

### **Definition**
#### SentencePiece is a modern subword tokenization algorithm and an open-source library developed by Google. It tokenizes text into subwords without requiring whitespace between words.


### **Why is SentencePiece Needed?**
- Handles unknown (Out-of-Vocabulary) words.
- Works with languages that do not use spaces.
- Reduces vocabulary size.
- Supports multilingual NLP.
- Improves Transformer and LLM performance.

### **How Does SentencePiece Work?**
1. Reads raw text directly.
2. Does not require pre-tokenization.
3. Learns a subword vocabulary.
4. Splits text into meaningful subword units.
5. Uses either **BPE** or **Unigram Language Model**.

### **Key Features**
- Subword-based tokenization.
- No whitespace required.
- Supports multiple languages.
- Handles unknown words.
- Smaller vocabulary.

### **Advantages**
- Works without spaces.
- Excellent multilingual support.
- Better handling of rare words.
- Easy to train on custom datasets.

### **Limitations**
- Requires vocabulary training.
- More complex than word tokenization.
- Needs a trained SentencePiece model.

### **Applications**
- T5
- ALBERT
- XLNet
- mT5
- LLaMA
- Gemma
- Mistral

### **Algorithms Used**
- Byte Pair Encoding (BPE)
- Unigram Language Model

````python
# Install Library 
#!pip install sentencepiece transformers
````

````python
# Import Library
from transformers import AutoTokenizer
from collections import Counter

# Load T5 SentencePiece Tokenizer
tokenizer = AutoTokenizer.from_pretrained("t5-small")

# Input Text
text = "Machine Learning is transforming the world."

# Tokenize Text
tokens = tokenizer.tokenize(text)

# Print Tokens
print("Tokens:")
print(tokens)

# Total Tokens
print("\nTotal Tokens:", len(tokens))
````

````python
# Print Token Number
print("\nToken Number:")
for i, token in enumerate(tokens, start=1):
    print(f"Token {i}: {token}")
````

````python
# Encode Text into Token IDs
token_ids = tokenizer.encode(text)

print("\nToken IDs:")
print(token_ids)
````

````python
# Decode Token IDs
print("\nDecoded Text:")
print(tokenizer.decode(token_ids))
````

````python
# Convert Tokens to IDs
print("\nTokens to IDs:")
token_ids = tokenizer.convert_tokens_to_ids(tokens)
print(token_ids)
````

````python
# Convert IDs to Tokens
print("\nIDs to Tokens:")
print(tokenizer.convert_ids_to_tokens(token_ids))
````

````python
# Vocabulary Size
print("\nVocabulary Size:", tokenizer.vocab_size)
````

````python
# Unique Tokens
print("\nUnique Tokens:", len(set(tokens)))
````

````python
# Token Frequency
frequency = Counter(tokens)

print("\nToken Frequency:")
print(frequency)
````

````python
# Top 10 Frequent Tokens
print("\nTop 10 Frequent Tokens:")
print(frequency.most_common(10))
````

````python
# Check Data Type
print("\nData Type:")
print(type(tokens))
````

# Interview Questions & Answers – Modern Tokenization

### Q1. What is Modern Tokenization?
**Answer:** Modern Tokenization is an advanced tokenization technique that splits text into **subwords** instead of complete words. It helps NLP models handle unknown words, reduce vocabulary size, and improve language understanding.

---

### Q2. Why is Modern Tokenization needed?
**Answer:**
- Handles unknown (OOV) words.
- Reduces vocabulary size.
- Represents rare words efficiently.
- Improves Transformer and LLM performance.

---

### Q3. What is Subword Tokenization?
**Answer:** Subword Tokenization splits words into smaller meaningful units called **subwords**, allowing models to handle unseen words efficiently.

---

### Q4. Why is Subword Tokenization better than Word Tokenization?
**Answer:**
- Handles unknown words.
- Smaller vocabulary.
- Better multilingual support.
- Higher model accuracy.

---

### Q5. What is Byte Pair Encoding (BPE)?
**Answer:** BPE is a frequency-based subword tokenization algorithm that repeatedly merges the most frequent pair of characters or subwords to build a vocabulary.

**Answer:**
- BERT
- DistilBERT
- ELECTRA

---

### Q9. What is SentencePiece?

### Q10. Which models use SentencePiece?
